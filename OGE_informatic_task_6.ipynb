```python
# 4-ый тип задания. // - Деление на цело
# Ниже приведена программа, записанная на языке программирования python

s = int(input())
k = int(input())
if s // 2 == k:
    print("ДА")
else:
    print("НЕТ")

# // - это операнд деления нацело
# Было проведено 9 запусков программы, при которых в качестве значений переменных s и t:
# (1, 1); (8, 4); (14, 10); (20, 1); (7, 3); (10, 5); (10, 2); (4, 1); (1, 0).
# Сколько было запусков, при которых программа напечатала «ДА»?
```

```python
#РЕШЕНИЕ
```

```python
# Программа для проверки
s = int()
k = int()
test_num = 1
main_kolvo = int()
input_matrix = [[1,1], [8,4], [14,10], [20,1], [7,3], [10,5], [10,2], [4,1], [1,0]]
for i in range(len(input_matrix)):
    for j in range(1):
        s = input_matrix[i][j]
        k = input_matrix[i][j+1]
        if s // 2 == k:
            print(f" {test_num}) {s, t} - ДА")
            main_kolvo += 1
        else:
            print((f" {test_num}) {s, t} - НЕТ"))
    test_num += 1
print("-----------------------------------------------")
print(f"Итоговое число вывод программой слова 'ДА' = {main_kolvo}")
print(f"Ответ: {main_kolvo}")
```

```python
# Решение
```

```python
# Программа для проверки

s = int()
k = int()
test_num = 1
main_kolvo = int()
input_matrix = [[1,2], [8,4], [6,-12], [-5,-5], [3,11], 
                [-10,12], [-10,-2], [4,1], [2,5]]
for i in range(len(input_matrix)):
    for j in range(1):
        s = input_matrix[i][j]
        k = input_matrix[i][j+1]
    if s >= 2 * k:
        print(f" {test_num}) {s, k} - ДА")
        main_kolvo += 1
    else:
        print(f" {test_num}) {s, k} - НЕТ")
    test_num += 1
print("-----------------------------------------------")
print(f"Итоговое число вывод программой слова 'ДА' = {main_kolvo}")
print(f"Ответ: {main_kolvo}")
```

```python
# 6-ой тип задания. if not - если не
# Ниже приведена программа, записанная на языке программирования python

s = int(input())
t = int(input())
if not ((s < -4) and (t >= 3)):
    print("YES")
else:
    print("NO")

# Было проведено 9 запусков программы, при которых в качестве значений переменных s и t:
# (−5, −2); (5, 3); (−14, 3); (−12, 5); (5, –7); (10, 3); (−4, 3); (3, 0); (−4, 9).
# Сколько было запусков, при которых программа напечатала «NO»?
```

```python
# РЕШЕНИЕ
```

```python
# Программа для проверки

s = int()
t = int()
test_num = 1
main_kolvo = int()
input_matrix = [[-5,-2], [5,3], [-14,3], [-12,5], [5,-7], 
                [10,3], [-4,3], [3,0], [-4,9]]
for i in range(len(input_matrix)):
    for j in range(1):
        s = input_matrix[i][j]
        t = input_matrix[i][j+1]
        if not ((s < -4) and (t >= 3)):
            print(f" {test_num}) {s, t} - YES")
        else:
            print(f" {test_num}) {s, t} - NO")
            main_kolvo += 1
    test_num += 1
print("-----------------------------------------------")
print(f"Итоговое число вывод программой слова 'NO' = {main_kolvo}")
print(f"Ответ: {main_kolvo}")      
        
```

```python
# 7-ой тип задания. % - деление с остатком
# Ниже приведена программа, записанная на языке программирования python

s = int(input())
 k = int(input())
if s % 5 == k:
    print("YES")
else:
    print("NO")

# Было проведено 9 запусков программы, при которых в качестве значений переменных s и t:
# (1, 2); (5, 0); (18, 3); (21; 1); (10, 2); (12, 1); (9; 4); (5, 5); (3, 3)
# Сколько было запусков, при которых программа напечатала «YES»?
```

```python
# РЕШЕНИЕ 
# % - это операнд деления с остатком
```

```python
# Программа для проверки
s = int()
k = int()
test_num = 1
main_kolvo = int()
input_matrix = [[1,2], [5,0], [18,3], [21,1], [10,2], 
                [12,1], [9,4], [5,5], [3,3]]
for i in range(len(input_matrix)):
    for j in range(1):
        s = input_matrix[i][j]
        k = input_matrix[i][j+1]
        if s % 5 == k:
            print(f" {test_num}) {s, t} - YES")
            main_kolvo += 1
        else:
            print(f" {test_num}) {s, t} - NO")            
    test_num += 1
print("-----------------------------------------------")
print(f"Итоговое число вывод программой слова 'YES' = {main_kolvo}")
print(f"Ответ: {main_kolvo}")
```
