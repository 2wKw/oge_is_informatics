```python
# Восстановите IP-адрес. В ответе укажите последовательность букв, обозначающих 
# фрагменты, в порядке, соответствующем IP-адресу.
```

```python
# Теория
# IP-адрес представляет собой четыре разделенных точками числа, каждое из которых не может
# быть болье 255

# Решение
# Рассмотри фрагмент "В" =  '.65' - так как по определению числа в IP-адресе не могут быть 
# больше 255 => мы не можем добавить в конце этого числа ещё один разряд => 
# данный фрагмент должен быть последним.
# Далее расммотрим фрагмент "Г" = "8.121" - К этому фрагменту также нельзя добавить ещё
# один разряд в конце => после данного фрагмента должен идти фрагмент начинающийся с
# точки => после фрагмента "Г" должен идти фрагмент "В". Таким образом фрагмент "Г"
# будет предпоследним
# Первым будет фрагмент Б => вторым будет фрагмент А
# => восстановленный i-адресс: 162.178.121.65
# и правильный ответ: БАГВ
```
